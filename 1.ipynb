```python
import tensorflow as tf
from tensorflow.keras import layers, models
import numpy as np 
import matplotlib.pyplot as plt
from tensorflow.keras.utils import load_img, img_to_array

tf.random.set_seed(42)

dataset = tf.keras.utils.image_dataset_from_directory(
    "datos",
    image_size = (64,64),
    batch_size = 8,
    label_mode = "binary"
)

print("clases detectadas", dataset.class_names)
```

```python
# Modelo Convolucional (CNN) optimizado para visión artificial
# Las capas Conv2D + MaxPooling permiten extraer bordes, texturas y formas de animales
model_denso = models.Sequential([
    layers.Input(shape=(64, 64, 3)),
    layers.Rescaling(1./255),
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.1),
    layers.Conv2D(32, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    layers.Conv2D(64, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    layers.Conv2D(128, (3, 3), activation='relu'),
    layers.MaxPooling2D(2, 2),
    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(32, activation='relu'),
    layers.Dense(1, activation='sigmoid')
], name='Red_Convolucional')

model_denso.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0005),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model_denso.summary()

historial_denso = model_denso.fit(dataset, epochs=30)
```

```python
def interpretar(valor):
    if valor > 0.5:
        return f'PERRO ({valor*100:.1f}%)'
    else:
        return f'GATO ({(1-valor)*100:.1f}%)'
```

```python
from numpy.testing import verbose
model_denso.save("modelo_denso.keras")
ruta_prueba="datos/golden-retriever.webp"
img = load_img(ruta_prueba, target_size=(64,64))
x = np.expand_dims(img_to_array(img), axis=0)

model_cargado = tf.keras.models.load_model("modelo_denso.keras")

prend = model_cargado.predict(x, verbose=0)[0][0]

print(f"Resultado {interpretar(prend)}")
print(prend)
```
